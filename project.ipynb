# Float convert into integer 

```python
def float_into_int(lis):
    for i in lis:
       df[i]= df[i].astype("int64")
    
fl=['age', 'bmi', 'HbA1c_level']
float_into_int(fl)
```

# Import label encoder from sklearn for convert object into integer

```python
from sklearn.preprocessing import LabelEncoder
```

```python
def object_into_int(lis):
    for i in lis:
        data =pd.DataFrame({i:df[i].unique()})
        data_le=LabelEncoder()
        data_le.fit(np.ravel(data))
        df[i]=data_le.transform(df[i])
obj=["gender","smoking_history"]
object_into_int(obj)
```

```python
df.info()
```

```python
X = df.drop('diabetes', axis=1)   # Dependant columns / Input columns
y = df['diabetes']                # Resultant columns / Output column
```

```python
from sklearn.model_selection import train_test_split
train_X, test_X, train_y, test_y = train_test_split(X,y, test_size=0.3, shuffle=False) # 80:20, 70:30, 60:40
```

```python
print(train_X.shape, train_y.shape)
print(test_X.shape, test_y.shape)
```

```python
from sklearn.svm import SVC # Support Vector Classifier

model_svc = SVC()
model_svc.fit(train_X, train_y)

print(model_svc)
```

```python
model_pred_svc = model_svc.predict(test_X)
model_pred_svc
```

```python
from sklearn.metrics import accuracy_score
model_acc_svc = accuracy_score(test_y, model_pred_svc)
print(round(model_acc_svc,3)*100, "%")
```

```python
import pickle
pickle.dump(model_svc, open('model_svc.pkl', 'wb'))   # Model Saving
```
